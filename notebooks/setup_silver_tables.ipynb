### Create silver price-paid transactions table

```python
from landreg.config import SILVER_SCHEMA, table

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {table(SILVER_SCHEMA, "price_paid_transactions")} (
        transaction_id STRING NOT NULL,
        price BIGINT NOT NULL,
        date_of_transfer DATE NOT NULL,
        transfer_year INT GENERATED ALWAYS AS (YEAR(date_of_transfer)),
        postcode STRING,
        property_type STRING,
        old_or_new STRING,
        duration STRING,
        paon STRING,
        saon STRING,
        street STRING,
        locality STRING,
        town_city STRING,
        district STRING,
        county STRING,
        category_type STRING,
        is_deleted BOOLEAN NOT NULL,
        is_price_outlier BOOLEAN NOT NULL,
        has_unknown_category BOOLEAN NOT NULL,
        run_id STRING NOT NULL,
        first_seen TIMESTAMP NOT NULL,
        last_seen TIMESTAMP NOT NULL
    ) PARTITIONED BY (transfer_year)
""")

```

### Create silver quarantine table

```python
from landreg.config import SILVER_SCHEMA, table

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {table(SILVER_SCHEMA, "price_paid_quarantine")} (
        transaction_id STRING,
        price STRING,
        date_of_transfer STRING,
        postcode STRING,
        property_type STRING,
        old_or_new STRING,
        duration STRING,
        paon STRING,
        saon STRING,
        street STRING,
        locality STRING,
        town_city STRING,
        district STRING,
        county STRING,
        category_type STRING,
        record_status STRING,
        run_id STRING NOT NULL,
        source_filename STRING NOT NULL,
        ingestion_timestamp TIMESTAMP NOT NULL,
        quarantine_reason STRING NOT NULL,
        quarantined_at TIMESTAMP NOT NULL
    )
""")
```
