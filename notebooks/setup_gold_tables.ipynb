### Create property dimension table

```python
from landreg.config import GOLD_SCHEMA, table

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {table(GOLD_SCHEMA, "dim_property")} (
        property_sk BIGINT NOT NULL,
        property_nk STRING NOT NULL,
        postcode STRING,
        property_type STRING,
        duration STRING,
        paon STRING,
        saon STRING,
        street STRING,
        locality STRING,
        town_city STRING,
        district STRING,
        county STRING,
        valid_from DATE NOT NULL,
        valid_to DATE,
        is_current BOOLEAN NOT NULL
    )
""")
```

### Create date dimension table

```python
from landreg.config import GOLD_SCHEMA, table

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {table(GOLD_SCHEMA, "dim_date")} (
        date_key INT NOT NULL,
        full_date DATE NOT NULL,
        day_number INT NOT NULL,
        day_name STRING NOT NULL,
        day_of_week INT NOT NULL,
        month_number INT NOT NULL,
        month_name STRING NOT NULL,
        calendar_quarter INT NOT NULL,
        calendar_year INT NOT NULL,
        fiscal_year INT NOT NULL,
        is_weekend BOOLEAN NOT NULL
    )
""")
```

### Create transaction fact table

```python
spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {table(GOLD_SCHEMA, "fact_transaction")} (
        transaction_id STRING NOT NULL,
        property_sk BIGINT NOT NULL,
        date_key INT NOT NULL,
        date_of_transfer DATE NOT NULL,
        transfer_year INT GENERATED ALWAYS AS (YEAR(date_of_transfer)),
        price BIGINT NOT NULL,
        old_or_new STRING,
        category_type STRING,
        is_deleted BOOLEAN NOT NULL,
        is_price_outlier BOOLEAN NOT NULL,
        has_unknown_category BOOLEAN NOT NULL,
        run_id STRING NOT NULL
    ) PARTITIONED BY (transfer_year)
""")
```
